```python
a=int(input('월을 입력하세요'))
if 0>=a or a>=13:
  print('입력이 잘못되었습니다')
elif 3<=a<=5:
  print('봄')
elif 6<=a<=8:
  print('여름')
elif 9<=a<=11:
  print('가을')
else:
  print('겨울')
```
